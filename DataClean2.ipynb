```python
import pandas as pd

df=pd.read_csv(
    "data/Perovskite_Data.csv", low_memory=False 
)
print(df.shape)
```

```python
#count the number of missing values in each column
miss_values = df.isna().sum()

#store columns with x missing values 
miss_columns = miss_values[miss_values > 30000]

#display the columns and how many values are missing in each
print(miss_columns)
```

```python
# Calculate the percentage of missing values
# only for the columns already stored in missing_columns
miss_percent = (
    df[miss_columns.index].isna().sum()
    / 42851
) * 100

# Display each selected column and its missing percentage
print(miss_percent)
```

```python
# Remove every column that contains at least one missing value
df_clean = df.drop(columns=miss_columns.index)

# Display the cleaned dataset
df_clean.head()
```

```python
df_unknown = df_clean.copy() 
df_unknown = df_unknown.replace("Unknown", pd.NA)
miss_values_unknown = df_unknown.isna().sum()

miss_columns_unknown=miss_values_unknown[miss_values_unknown > 30000]
print(miss_columns_unknown)
```
