```python
import pandas as pd

df=pd.read_csv(
    "data/Perovskite_database_content_all_data.csv", low_memory=False 
)
print(df.shape)
```

```python
display(df.head())
```

```python
print(df.columns.tolist())
```

```python
missing = df.isnull().sum()
display(
    missing[missing > 0].sort_values(ascending=False)
)
```

```python
df_clean = df.dropna(axis=1, how="all")
print("Original dimensions:", df.shape)
print("New dimensions:", df_clean.shape)
print("Columns removed:", df.shape[1]-df_clean.shape[1])
```

```python
print(df.isna().all().sum())
```

```python
keywords = [
    "Perovskite_composition",
    "bandgap",
    "band_gap",
    "JV_",
    "ETL",
    "HTL",
    "Substrate",
    "Backcontact",
    "Cell_architecture"
]
relevant_col = [
    col for col in df_clean.columns
    if any(word.lower() in col.lower() for word in keywords)
]
print(relevant_col)
```

```python

summary = pd.DataFrame({
    "available": df_clean[relevant_col].notna().sum(),
    "missing" : df_clean[relevant_col].isna().sum()
})
summary["missing %"] = (summary["missing"]/len(df_clean)*100)
display(summary)
```
